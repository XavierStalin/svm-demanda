# Probar los tres modelos mediante Run ID

Este cuaderno recupera los modelos guardados en MLflow y predice sin reentrenarlos.
Puedes usar tu kernel habitual: la función ejecuta automáticamente la RNN en
`.venv-rnn`, donde TensorFlow tiene versiones compatibles de NumPy y protobuf.
No cambia los paquetes del kernel ni requiere que la interfaz web esté abierta.

Solo se eligen automáticamente comparaciones de tres modelos que ya terminaron.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display
candidatas = [Path.cwd(), *Path.cwd().parents,
              Path.cwd() / "Repositorio TrABAJO U2" / "svm-demanda"]
base = next(p for p in candidatas if (p / "prueba_modelos_mlflow.py").is_file())
if str(base) not in sys.path: sys.path.insert(0, str(base))
import importlib
import prueba_modelos_mlflow
importlib.reload(prueba_modelos_mlflow)
from prueba_modelos_mlflow import probar_modelo_desde_mlflow, ids_ultima_comparacion
print("Python del cuaderno:", sys.executable)
print("RNN: carga automática en .venv-rnn cuando sea necesario.")
ids = ids_ultima_comparacion()
display(pd.Series(ids, name="Run ID").to_frame())
```

## Misma función, distinto ID

Cambiar `RNN` por `SVR inicial` o `SVR optimizada`. También puedes pegar un Run ID
literal en `ID_ELEGIDO`. Se usan los siete días anteriores a cada fecha y se informa
la venta real solo para calcular el error. Todas las cantidades están en miles de unidades.
Los IDs de candidatas sin modelo guardado no permiten hacer esta prueba.

```python
ID_ELEGIDO = ids["SVR optimizada"] #"RNN" por "SVR inicial" o "SVR optimizada"
fechas = ["2016-09-13", "2017-08-15"]
resultado = probar_modelo_desde_mlflow(ID_ELEGIDO, fechas)
display(resultado.round(3))
```

## Comparar los tres con las mismas fechas

```python
comparacion_por_id = pd.concat([
    probar_modelo_desde_mlflow(run_id, fechas) for run_id in ids.values()
], ignore_index=True)
display(comparacion_por_id.round(3))
```
